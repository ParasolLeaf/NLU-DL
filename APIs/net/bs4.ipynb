```python
import requests
from bs4 import BeautifulSoup
import re

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/68.0.3440.106 Safari/537.36',
}

response = requests.get('https://www.baidu.com/s?ie=utf-8&tn=baidu&wd=北京大学',headers=headers)
print(response.status_code)  # 打印状态码
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, 'html.parser')

div_contents = soup.find("div", id="content_left")

links = []
for a_tag in div_contents.find_all('a', href=True):
    href = a_tag['href']
    text = a_tag.text.strip()
    try:
        r = requests.get(href, headers=headers, timeout=10)
        print("真实链接：", r.url, text)
        if href.startswith('http') or href.startswith('https'):
            links.append(r.url)
    except:
        continue
print(links)
```
